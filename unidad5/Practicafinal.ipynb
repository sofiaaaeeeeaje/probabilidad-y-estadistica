```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from scipy.stats import pearsonr
from IPython.display import display
from google.colab import files
uploaded = files.upload()

sns.set(style="whitegrid")

def ecuacion(a, b, dec=2):
    signo = "+" if b >= 0 else "-"
    return f"Ŷ = {a:.{dec}f} {signo} {abs(b):.{dec}f}X"

def analizar_grupo(n):
    print(f"Grupo {n}")

    print(f"\n{n}.1 Carga y exploración inicial\n")
    df = pd.read_csv(f"grupo{n}.csv")
    display(df.head())

    print("\nFilas y columnas:", df.shape, "\n")
    df.info()
    print("\nValores nulos por columna:")
    print(df.isnull().sum())

    print("\nEstadísticos descriptivos:")
    display(df.describe())

    plt.figure(figsize=(7, 4))
    sns.histplot(df["Calificacion"], bins=10, kde=True, color="steelblue")
    plt.title(f"Histograma de Calificaciones Finales - Grupo {n}")
    plt.xlabel("Calificación Final")
    plt.ylabel("Frecuencia")
    plt.show()

    plt.figure(figsize=(7, 4))
    sns.scatterplot(x="HorasEstudio", y="Calificacion", data=df, color="blue")
    plt.title(f"Diagrama de Dispersión: Horas de Estudio vs Calificación - Grupo {n}")
    plt.xlabel("Horas de Estudio")
    plt.ylabel("Calificación Final")
    plt.grid(True)
    plt.show()

    print(f"\n{n}.2 Modelo de regresión lineal simple\n")
    X = df[["HorasEstudio"]]
    y = df["Calificacion"]

    modelo = LinearRegression()
    modelo.fit(X, y)

    b = modelo.coef_[0]
    a = modelo.intercept_
    r2_modelo = modelo.score(X, y)

    print(f"Pendiente (b): {b:.4f}")
    print(f"Intercepto (a): {a:.4f}")
    print(f"R² del modelo: {r2_modelo:.4f}")
    print(f"Ecuación de regresión: {ecuacion(a, b, 4)}")

    horas = 10
    pred = modelo.predict(pd.DataFrame({"HorasEstudio": [horas]}))
    print(f"Calificación estimada con {horas} horas de estudio: {pred[0]:.2f}")

    print(f"\n{n}.3 Correlación de Pearson\n")
    r, p = pearsonr(df["HorasEstudio"], df["Calificacion"])
    print(f"Coeficiente de correlación de Pearson (r): {r:.4f}")
    print(f"Valor p: {p:.4e}")
    print(f"Coeficiente de determinación (r²): {r**2:.4f}")

    print(f"\n{n}.4 Visualización: datos y recta de regresión\n")
    y_pred = modelo.predict(X)

    plt.figure(figsize=(8, 5))
    plt.scatter(df["HorasEstudio"], df["Calificacion"], color="blue", label="Datos")
    plt.plot(df["HorasEstudio"], y_pred, color="red",
             label=f"Línea de regresión: {ecuacion(a, b)}")
    plt.title(f"Regresión Lineal: Horas de Estudio vs Calificación Final - Grupo {n}")
    plt.xlabel("Horas de Estudio")
    plt.ylabel("Calificación Final")
    plt.legend(title=f"r = {r:.3f}")
    plt.grid(True)
    plt.show()

    plt.figure(figsize=(8, 5))
    sns.regplot(x="HorasEstudio", y="Calificacion", data=df,
                scatter_kws={"color": "blue"}, line_kws={"color": "red"})
    plt.title(f"Diagrama de Dispersión con Línea de Regresión (seaborn) - Grupo {n}")
    plt.xlabel("Horas de Estudio")
    plt.ylabel("Calificación Final")
    plt.show()

    return df, modelo, X, a, b, r, p

df1, modelo1, X1, a1, b1, r1, p1 = analizar_grupo(1)
df2, modelo2, X2, a2, b2, r2, p2 = analizar_grupo(2)
df3, modelo3, X3, a3, b3, r3, p3 = analizar_grupo(3)

print("Resumen comparativo\n")

resumen = pd.DataFrame({
    "Grupo": ["Grupo 1", "Grupo 2", "Grupo 3"],
    "Intercepto (a)": [a1, a2, a3],
    "Pendiente (b)": [b1, b2, b3],
    "r (Pearson)": [r1, r2, r3],
    "r²": [r1**2, r2**2, r3**2],
}).round(4)
resumen["Valor p"] = [f"{p:.2e}" for p in (p1, p2, p3)]
resumen["Ecuación"] = [ecuacion(a1, b1), ecuacion(a2, b2), ecuacion(a3, b3)]
display(resumen)

fig, ejes = plt.subplots(1, 3, figsize=(15, 4.5), sharey=True)
datos = [(df1, modelo1, X1, "Grupo 1", r1),
         (df2, modelo2, X2, "Grupo 2", r2),
         (df3, modelo3, X3, "Grupo 3", r3)]

for eje, (df, modelo, X, nombre, r) in zip(ejes, datos):
    eje.scatter(df["HorasEstudio"], df["Calificacion"], color="blue", label="Datos")
    eje.plot(df["HorasEstudio"], modelo.predict(X), color="red", label="Línea de regresión")
    eje.set_title(f"{nombre} (r = {r:.3f})")
    eje.set_xlabel("Horas de Estudio")
    eje.grid(True)
    eje.legend()

ejes[0].set_ylabel("Calificación Final")
plt.suptitle("Comparación de la regresión lineal en los tres grupos")
plt.tight_layout()
plt.show()
```
